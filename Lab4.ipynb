```python
#Detecting Outliers in House Price

Q1 = df['House_Price'].quantile(0.25) 
Q3 = df['House_Price'].quantile(0.75) 
IQR = Q3 - Q1 
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['House_Price'] < lower_bound) | (df['House_Price'] > upper_bound)]
outliers.head(15)
```

```python
#Detecting Outliers in Square Footage
Q1 = df['Square_Footage'].quantile(0.25) 
Q3 = df['Square_Footage'].quantile(0.75) 

IQR = Q3 - Q1 
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['Square_Footage'] < lower_bound) | (df['Square_Footage'] > upper_bound)]
outliers.head(15)
```

```python
# Detecting Outliers in lot size 

Q1 = df['Lot_Size'].quantile(0.25) 
Q3 = df['Lot_Size'].quantile(0.75) 

IQR = Q3 - Q1 
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['Lot_Size'] < lower_bound) | (df['Lot_Size'] > upper_bound)]
outliers.head(15)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df_scaled = df[["Square_Footage", "Lot_Size","Num_Bedrooms","Num_Bathrooms","Garage_Size","Neighborhood_Quality","Year_Built"]].copy()

df_scaled[["Square_Footage", "Lot_Size","Num_Bedrooms","Num_Bathrooms","Garage_Size","Neighborhood_Quality","Year_Built"]] = scaler.fit_transform(df_scaled[["Square_Footage", "Lot_Size","Num_Bedrooms","Num_Bathrooms","Garage_Size","Neighborhood_Quality","Year_Built"]])

df_scaled.head()

```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_standardized = df[["Square_Footage", "Lot_Size","Num_Bedrooms","Num_Bathrooms","Garage_Size","Neighborhood_Quality","Year_Built"]].copy()

df_standardized[["Square_Footage", "Lot_Size","Num_Bedrooms","Num_Bathrooms","Garage_Size","Neighborhood_Quality","Year_Built"]] = scaler.fit_transform(df_standardized[["Square_Footage", "Lot_Size","Num_Bedrooms","Num_Bathrooms","Garage_Size","Neighborhood_Quality","Year_Built"]])

df_standardized.head()

```

```python
plt.figure(figsize=(8,5)) 
sns.heatmap(df_standardized.corr(),annot=True, cmap="coolwarm")
plt.title("Correlation Matrix of Standardized Features")
plt.show()
```

```python
from sklearn.decomposition import PCA 

X = df_standardized[["Square_Footage", "Lot_Size","Num_Bedrooms","Num_Bathrooms","Garage_Size","Neighborhood_Quality","Year_Built"]] 

pca = PCA()

principal_components = pca.fit_transform(X)

print("Explained Variance Ratio:", pca.explained_variance_ratio_)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(principal_components[:, 0], principal_components[:, 1], c=df['House_Price'])
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.colorbar(label='House Price')
plt.title('Houses Projecting onto PC Space (Colored by Price)')
plt.show()
```

# Task5 

The PCA results indicate that the variance is spread thinly across all components. Because the first two components capture less than 50% of the total variance, it suggests the features are largely independent. We cannot perform extreme dimensionality reduction without losing a significant amount of the dataset's information.
