```python
#Lab3
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns   

#load the dataset
df = pd.read_csv('house_price_regression_dataset.csv')


#print first 5 rows
print("first rows: \n",df.head())

#column names and data types
print("\n column name and data types: \n",df.dtypes)

#check for missing values 
print("\n missing values: \n",df.isnull())
print(df.isna().sum())

#print duplicated rows
df.duplicated().sum()[df.duplicated == True]

#print the number of rows and columns
print("Shape: \n",df.shape)

print("number of rows: \n",df.shape[0])
print("number of columns: \n",df.shape[1])

#statistical summary of the dataset
print("\n statistical summary: \n",df.describe(include='all'))

plt.figure(figsize=(10,6))
sns.histplot(df['House_Price'], bins=20)
plt.title('Distribution of House Prices')
plt.show()

plt.figure(figsize=(10,6))
sns.histplot(df['Lot_Size'],bins=20)
plt.title('Lot Size Distribution')
plt.show()
```
