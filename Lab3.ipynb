```python
#Lab3

#importing libraries
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns   

#make cleaner graphs
sns.set()





```

```python
#load the dataset
df = pd.read_csv('house_price_regression_dataset.csv')


#print first 5 rows
print("first rows: \n",df.head())
```

```python
# Check all missing values
print(df.isna())
```

```python
#number of missing values in each column
print(df.isna().sum())
```

```python
# checking duplicated rows
print("Duplicate rows count:", df.duplicated().sum())

# If you want to see the actual duplicate rows (if any)
print(df[df.duplicated()])
```

```python
#print the number of rows and columns
print("Shape: \n",df.shape)
print("number of rows: \n",df.shape[0])
print("number of columns: \n",df.shape[1])
```

```python
#column names and data types
print("\n column name and data types: \n",df.dtypes)
```

```python
#statistical summary of the dataset
print("\n statistical summary: \n",df.describe(include='all'))
```

```python
# Distribution of House Price
plt.figure(figsize=(8,5))
sns.histplot(df['House_Price'], bins=20,)
plt.title("Distribution of House Price")
plt.show()
```

```python
# Distribution of Square Footage
plt.figure(figsize=(8,5))
sns.histplot(df['Square_Footage'], bins=20)
plt.title("Distribution of Square Footage")
plt.show()
```

```python
# Average Price by Number of Bedrooms
plt.figure(figsize=(10,5))
df.groupby('Num_Bedrooms')['House_Price'].mean().plot(kind='bar', color='orange')
plt.title("Average House Price by Number of Bedrooms")
plt.ylabel("Average Price")
plt.show()
```

```python
# Average Price by Garage Size
plt.figure(figsize=(10,5))
df.groupby('Garage_Size')['House_Price'].mean().plot(kind='bar', color='green')
plt.title("Impact of Garage Size on House Price")
plt.ylabel("Average Price")
plt.show()
```

```python
# Average Price by Neighborhood Quality
plt.figure(figsize=(10,5))
neigh_price = df.groupby('Neighborhood_Quality')['House_Price'].mean()
neigh_price.plot(kind='bar', color='skyblue')
plt.title("Average House Price by Neighborhood Quality")
plt.ylabel("Average Price")
plt.show()
```

```python
# Square Footage vs House Price Relationship
plt.figure(figsize=(8,5))
sns.scatterplot(x='Square_Footage', y='House_Price', data=df)
plt.title("Square Footage vs House Price")
plt.show()
```

```python
# Correlation Matrix
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

```python
# Average Price Trend by Year Built
plt.figure(figsize=(10,5))
year_trend = df.groupby('Year_Built')['House_Price'].mean()
year_trend.plot()
plt.title("Average House Price Trend by Year Built")
plt.ylabel("Average Price")
plt.xlabel("Year Built")
plt.show()
```
